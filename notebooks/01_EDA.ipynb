```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
df = pd.read_csv(r"../data/raw/eclipse_bug_report_data.csv")
df.drop(['assignee_name',
       'reporter_name', 'resolution_category', 'resolution_code',
       'status_category', 'status_code', 'update_date'],axis = 1,inplace=True)
```

```python
df["long_description"] = df["long_description"].fillna("")

df["text"] = df["short_description"] + " " + df["long_description"]
```

```python
df.duplicated().sum()
df["long_description"] = df["long_description"].fillna("")
df["text"] = (
    df["short_description"] + " " + df["long_description"]
)
df.drop(
    columns=[
        "bug_id",
        "short_description",
        "long_description",
        "severity_code"
    ],
    inplace=True,
)
df["creation_date"] = pd.to_datetime(df["creation_date"])
df["resolution_date"] = pd.to_datetime(df["resolution_date"])
df = df[df["bug_fix_time"] <= 365]
df.to_csv("../data/processed/processed_bugs.csv", index=False)
```
